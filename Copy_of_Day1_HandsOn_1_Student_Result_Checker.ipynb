## Introduction

Every useful program follows a simple pattern:

**Input --> Store Data --> Process --> Make Decision --> Display Result**

In this notebook you will apply each part of that pattern to build a working Student Result Checker from scratch.

## Scenario

You are creating a simple college **Student Result Checker**.

The application should:

1. Ask for the student's name.
2. Ask for marks.
3. Store the information.
4. Check the result.
5. Display the final result.

---

## Step 1 -- Collect Student Information

The `input()` function lets your program ask the user to type something.

- `input()` always returns a **string** (text).
- To work with numbers, wrap the input in `int()` to convert the text into an integer.

Run the cell below and type your responses when prompted.

````python
# Collect student name and marks
name = input("Enter student name: ")
marks = int(input("Enter marks: "))

# Display the stored values
print("Student:", name)
print("Marks:", marks)
````

**Expected Output** (if the user enters `Asha` and `82`):

```
Student: Asha
Marks: 82
```

Notice that `name` is a **string** and `marks` is an **integer**. Python stores each value with the correct data type so you can use it later.

---

## Step 2 -- Apply a Condition

Now we need the program to make a decision: did the student pass or fail?

- `if` checks whether a condition is `True`.
- `else` runs when the condition is `False`.

The comparison operator `>=` means "greater than or equal to".

````python
# Determine pass or fail
if marks >= 50:
    print("Result: Pass")
else:
    print("Result: Fail")
````

**How it works:**

- Python evaluates `marks >= 50`.
- If the expression is `True`, the indented code under `if` runs.
- If it is `False`, the indented code under `else` runs.

Only **one** branch executes -- never both.

---

## Step 3 -- Add Result Categories

A simple pass/fail is not enough. Let us add a third category: **Distinction** for students who score 75 or above.

`elif` (short for "else if") lets you check additional conditions between `if` and `else`.

````python
# Categorize the result
if marks >= 75:
    result = "Distinction"
elif marks >= 50:
    result = "Pass"
else:
    result = "Fail"

print("Result:", result)
````

**Expected Output** (if marks = 82):

```
Result: Distinction
```

Python checks the conditions from **top to bottom** and stops at the first one that is `True`.

---

## Step 4 -- Display the Final Result

Combine everything into a clean final display showing the student's name, marks, and result.

````python
# Display the complete student report
print("===== Student Result =====")
print("Name  :", name)
print("Marks :", marks)
print("Result:", result)
print("=========================")
````

**Expected Output** (if name = Asha, marks = 82):

```
===== Student Result =====
Name  : Asha
Marks : 82
Result: Distinction
=========================
```

## Final Challenge

Extend the Student Result Checker to handle **3 subjects**.

Your program should:

1. Accept the student's name.
2. Accept marks for **3 subjects** (e.g., Math, Science, English).
3. Calculate the **average** of the 3 marks.
4. Determine the result based on the average:
   - Average >= 75 --> Distinction
   - Average >= 50 --> Pass
   - Average < 50  --> Fail
5. Print a **formatted student report** showing each subject, the average, and the result.

Write your solution in the cell below.

````python
# --- Your solution here ---

# Collect student name and marks
name = input("Enter student name: ")

m = int(input("Enter sub1 mark: "))
n = int(input("Enter sub2 mark: "))
o = int(input("Enter sub3 mark: "))
avg = (m+n+o)/3
if(avg >= 75):
    result = "Distinction"
elif(avg >= 50):
    result = "Pass"
else:
    result = "Fail"
print("===== Student Result =====")
print("Name  :", name)
print("Sub1  :", m)
print("Sub2  :", n)
print("Sub3  :", o)
print("Avg   :", avg)
print("Result:", result)
print("=========================")
````
