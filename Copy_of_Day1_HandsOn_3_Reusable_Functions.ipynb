# Hands-on 3 -- Make Your Python Code Reusable

## Objective

Convert the Student Result Checker into reusable functions using:

- Function definitions (`def`)
- Parameters
- Arguments
- Return values

## Introduction

In Hands-on 1 the entire program was written as one long sequence of code. That approach works for small scripts, but as programs grow it becomes hard to read, test, and maintain.

The solution is to break the program into **functions** -- small, named blocks of code that each do one specific job.

**Repeated Code --> Functions --> Reusable Program**

## Scenario

The Student Result Checker from Hands-on 1 works, but all the logic is written as one long script. If you wanted to reuse the result-checking logic for 100 students, you would need to copy and paste the same code 100 times.

Your task is to reorganize the program into **smaller reusable functions**.

---

## Step 1 -- Create an Input Function

A function is created using the `def` keyword. The `return` statement sends a value back to the code that called the function.

This function collects a student's name and marks from the user.

````python
def get_student_data():
    """Ask the user for a student's name and marks, then return both values."""
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    return name, marks
````

**Key points:**

- `def get_student_data():` defines the function. The parentheses `()` are empty because this function does not need any outside information to do its job.
- `return name, marks` sends **two values** back to whoever calls the function.
- The function does **not run yet**. It is only stored in memory until you call it.

---

## Step 2 -- Create a Result Function

This function receives marks and returns the result category.

````python
def check_result(marks):
    """Determine the result category based on marks."""
    if marks >= 75:
        return "Distinction"
    elif marks >= 50:
        return "Pass"
    else:
        return "Fail"
````

**Key points:**

- `marks` inside the parentheses is a **parameter** -- a placeholder for the actual value the function will receive.
- When the function runs, it evaluates the conditions and returns the matching string.
- The function does **one job**: determine the result. It does not print anything or ask for input.

---

## Step 3 -- Create a Display Function

This function handles all the output formatting.

````python
def display_result(name, marks, result):
    """Print a formatted student report."""
    print("===== Student Result =====")
    print("Student:", name)
    print("Marks  :", marks)
    print("Result :", result)
    print("=========================")
````

This function takes three parameters (`name`, `marks`, `result`) and displays them. It does not return anything -- its only job is to print.

---

## Step 4 -- Connect the Functions

Now call each function in the correct order to run the complete program.

````python
# Step A: Collect data
name, marks = get_student_data()

# Step B: Determine result
result = check_result(marks)

# Step C: Display the report
display_result(name, marks, result)
````

**Expected Output** (if the user enters `Priya` and `91`):

```
===== Student Result =====
Student: Priya
Marks  : 91
Result : Distinction
=========================
```

**Execution order:**

1. `get_student_data()` runs first -- it collects the name and marks.
2. `check_result(marks)` runs next -- it uses the marks to determine the result.
3. `display_result(name, marks, result)` runs last -- it prints the formatted report.

Each function does one clear job. The main program simply connects them together.

## Parameter vs Argument

These two terms are related but different. A simple example will make the distinction clear.

````python
def greet(name):       # 'name' is a PARAMETER (placeholder in the definition)
    print("Hello", name)

greet("Asha")          # "Asha" is an ARGUMENT (actual value passed in)
````

**Expected Output:**

```
Hello Asha
```

| Term | Where it appears | Example |
|---|---|---|
| **Parameter** | In the function **definition** | `name` in `def greet(name):` |
| **Argument** | In the function **call** | `"Asha"` in `greet("Asha")` |

Think of a parameter as an empty box with a label. The argument is the actual item you put into that box when you use the function.

## Quick Test -- Using Functions Multiple Times

The power of functions is that you can call them as many times as you need without rewriting the logic.

````python
# Test check_result with different values
print(check_result(92))   # Distinction
print(check_result(55))   # Pass
print(check_result(30))   # Fail
````

**Expected Output:**

```
Distinction

Fail
```

The same function handles all three cases. You wrote the logic once and reused it three times.

## Final Challenge

Take the Student Data Processor from Hands-on 2 and convert it into a function-based program.

Create the following functions:

- `calculate_average(students)` -- accepts the students dictionary, returns the class average.
- `count_results(students)` -- accepts the students dictionary, returns the number of students who passed and the number who failed.
- `find_top_student(students)` -- accepts the students dictionary, returns the name and marks of the highest performer.
- `display_report(students)` -- calls the other functions and prints a complete class report.

Then call `display_report(students)` to run the entire program.

Write your solution in the cell below.

````python
# Starter data (same as Hands-on 2)
students = {
    "Asha": 82,
    "Rahul": 45,
    "Priya": 91,
    "Arun": 68,
    "Neha": 39
}

# --- Your functions here ---

def calculate_average(students):
    total_marks = 0

    for marks in students.values():
        total_marks += marks

    return total_marks / len(students)



def count_results(students):
    pass_count = 0
    fail_count = 0

    for marks in students.values():
        if marks >= 50:
            pass_count += 1
        else:
            fail_count += 1

    return pass_count, fail_count



def find_top_student(students):
    top_student = max(students, key=students.get)

    return top_student, students[top_student]

# --- Call display_report(students) here ---
def display_report(students):
    average = calculate_average(students)
    pass_count, fail_count = count_results(students)
    top_student, top_marks = find_top_student(students)

    print("Class Average:", average)
    print("Pass Count:", pass_count)
    print("Fail Count:", fail_count)
    print("Top Student:", top_student)
    print("Top Marks:", top_marks)


display_report(students)
````

## Key Takeaway

Functions make programs:

- **Reusable** -- write once, call many times.
- **Organized** -- each function does one clear job.
- **Easier to test** -- you can test each function independently.
- **Easier to maintain** -- changes in logic only need to be made in one place.
